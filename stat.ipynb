```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd

# Sample data
data = pd.DataFrame({
    'Height': [160, 170, 180],
    'Salary': [50000, 55000, 120000]
})

# Min-Max Scaling
min_max = MinMaxScaler()
data_minmax = min_max.fit_transform(data)

# Standard Scaling
standard = StandardScaler()
data_standard = standard.fit_transform(data)

print("Min-Max Scaled:\n", data_minmax)
print("Standard Scaled:\n", data_standard)
```

```python
import numpy as np
import scipy.stats as stats

data = [2, 3, 4, 5, 6, 7, 8, 9]
mean = np.mean(data)
sem = stats.sem(data)  # standard error of the mean

# 95% Confidence Interval
ci = stats.t.interval(confidence=0.95, df=len(data)-1, loc=mean, scale=sem)

print("Mean:", round(mean, 2))
print("95% Confidence Interval:", ci)
```

```python
import numpy as np
import pandas as pd

data = [10, 12, 11, 13, 12, 300]  # 300 is an outlier
df = pd.DataFrame(data, columns=["Value"])

Q1 = df["Value"].quantile(0.25)
Q3 = df["Value"].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df["Value"] < Q1 - 1.5 * IQR) | (df["Value"] > Q3 + 1.5 * IQR)]
print("Outliers:\n", outliers)
```
